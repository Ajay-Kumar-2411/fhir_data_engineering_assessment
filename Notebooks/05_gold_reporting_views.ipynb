## Create Reporting Views

The assignment requires final Warehouse views optimized for reporting and analytics. The following T-SQL cells should be run against `FHIR_Gold_Warehouse`.

**Important:** In Fabric, make `FHIR_Gold_Warehouse` the primary Warehouse for the T-SQL cells below.

```sql
CREATE OR ALTER VIEW dbo.vw_patient_summary
AS
SELECT
    patient_id,
    active,
    gender,
    birth_date,
    deceased,
    source_last_updated,
    effective_from,
    effective_to,
    is_current
FROM dbo.dim_patient
WHERE is_current = 1;
```

```sql
CREATE OR ALTER VIEW dbo.vw_encounter_activity
AS
SELECT
    encounter_id,
    patient_id,
    status,
    class_code,
    class_display,
    period_start,
    period_end,
    source_last_updated
FROM dbo.fact_encounter
WHERE is_current = 1;
```

```sql
CREATE OR ALTER VIEW dbo.vw_observation_summary
AS
SELECT
    observation_id,
    patient_id,
    status,
    code_system,
    code,
    code_display,
    value_quantity,
    value_unit,
    effective_datetime,
    source_last_updated
FROM dbo.fact_observation
WHERE is_current = 1;
```

```sql
CREATE OR ALTER VIEW dbo.vw_condition_summary
AS
SELECT
    condition_id,
    patient_id,
    clinical_status,
    verification_status,
    code_system,
    code,
    code_display,
    onset_date,
    source_last_updated
FROM dbo.fact_condition
WHERE is_current = 1;
```

## Final Gold Validation

Validate that the reporting views are available in the Gold Warehouse.

```sql
SELECT 'vw_patient_summary' AS view_name, COUNT(*) AS record_count FROM dbo.vw_patient_summary
UNION ALL
SELECT 'vw_encounter_activity', COUNT(*) FROM dbo.vw_encounter_activity
UNION ALL
SELECT 'vw_observation_summary', COUNT(*) FROM dbo.vw_observation_summary
UNION ALL
SELECT 'vw_condition_summary', COUNT(*) FROM dbo.vw_condition_summary;
```
